```python
import pandas as pd
import numpy as np


from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import chromadb

print("All libraries imported successfully!")
print(f"ChromaDB version: {chromadb.__version__}")
```

```python
documents=[
            "Python loops are used to execute a block of code repeatedly.",
        "SQL is used to query, insert, update and delete data from databases.",
        "Regression is a machine learning technique used to predict continuous values.",
        "Functions help organize reusable code and improve readability.",
        "ETL stands for Extract Transform Load and is used in data engineering."
]
query_keyword="code"

print("*" *60)
print(f"KEYWORD SEARCH for:'{query_keyword}'")
print("*" *60)

for i,doc in enumerate(documents):
  if query_keyword.lower() in doc.lower():
    print(f"FOUND  [doc_{i}]: {doc}")
  else:
    print(f"MISSED [doc_{i}]:{doc}")

print()
print("PROBLEM: doc_2 talks about 'Cars and trucks' - which ARE vehicles")
print("But keyword search MISSED it because it searched for the exact word 'vehicle'")
```

```python
print("Loading embedding model... (maytake 1-2 minutes on first run)")
model=SentenceTransformer("all-MiniLM-L6-v2")
print("model loaded successfully!")

print(f"Model products vectors of size:{model.get_sentence_embedding_dimension()}dimensions")
```

```python

documents = [
     "Python loops are used to execute a block of code repeatedly.",
        "SQL is used to query, insert, update and delete data from databases.",
        "Regression is a machine learning technique used to predict continuous values.",
        "Functions help organize reusable code and improve readability.",
        "ETL stands for Extract Transform Load and is used in data engineering."
]

query = "code"

print("*" * 60)
print(f"SEMANTIC SEARCH for: '{query}'")
print("*" * 60)


query_embedding = model.encode(query)
doc_embeddings = model.encode(documents)

for i, doc_embedding in enumerate(doc_embeddings):
    similarity = cosine_similarity(
        [query_embedding],
        [doc_embedding]
    )[0][0]


    if similarity > 0.3:
        print(f"FOUND  [doc_{i}]: {documents[i]}")
    else:
        print(f"MISSED [doc_{i}]: {documents[i]}")

print()
print("Semantic search understands meaning.")
print("Python loops are used to execute a block of code repeatedly, so doc_0 is FOUND.")
```

```python
chroma_client =chromadb.Client()
collection=chroma_client.get_or_create_collection("demo_notes")
print("ChromaDB client created (in-memory mode)")


print(f"Collection name:demo_notes ")
print(f"Documents in collection :{collection.count()}")
```

```python
sample_docs = [
    "Python loops are used to execute a block of code repeatedly.",
    "SQL is used to query, insert, update and delete data from databases.",
    "Regression is a machine learning technique used to predict continuous values.",
    "Functions help organize reusable code and improve readability.",
    "ETL stands for Extract Transform Load and is used in data engineering."
]

sample_ids = ["N1", "N2", "N3", "N4", "N5"]

sample_metadata = [
    {"subject": "Python", "topic": "Loops"},
    {"subject": "DBMS", "topic": "SQL"},
    {"subject": "Machine Learning", "topic": "Regression"},
    {"subject": "Python", "topic": "Functions"},
    {"subject": "ETL", "topic": "Data Cleaning"}
]

collection.add(
    documents=sample_docs,
    ids=sample_ids,
    metadatas=sample_metadata
)

print("Documents added to collection!")
print(f"Total documents now in collection: {collection.count()}")
```
